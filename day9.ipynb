```python
# recusions

def countdown(n):
    if n<=0:     #base case condition when stop recursion
        print("Done !")
    else:
        print(n)
        countdown(n-1)  # recursive case
countdown(9)        
```

```python
# factorial (recursive fu7nction)
def fact(m):
    if m==0 or m==1:
        return 1
    else:
       return m*fact(m-1)
        
result2=fact(5)     
print(result2)   
```

```python
#fibbonacci series


def fibb(p):
    if p<=1:
        return p
    else:
       return fibb(p-1)+fibb(p-2)
print(fibb (7))   # 5th number
```

```python
def summe(numms):
    if len(numms)==0:
        return 0
    else:
        return numms[0]+summe(numms[1:])


num_list=[1,2,3,4,5]  
print(summe(num_list))  
```
