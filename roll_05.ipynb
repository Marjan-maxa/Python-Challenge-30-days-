```python
# list insert, remove, pop,reverse,find index

fruits=["goava","apple","cherry","banana"]
print(type(fruits))
x=fruits.index("banana")
print("Index( of banana :",x)
fruits.insert(2,"papaya")
print("After Insert :",fruits)
fruits.sort()
print("After Sorted data :",fruits)
fruits.sort(reverse=True)
print("Ater decending sort :",fruits)
fruits.remove("apple")
print(fruits)
fruits.pop(1)
print(fruits)
```

```python
def myFunc(e):
  return e['year']

cars = [
  {'car': 'Ford', 'year': 2005},
  {'car': 'Mitsubishi', 'year': 2000},
  {'car': 'BMW', 'year': 2019},
  {'car': 'VW', 'year': 2011}
]

cars.sort(key=myFunc)

print(cars)
```

```python
students = [
    {"name": "Marjan", "marks": 85},
    {"name": "Rahim", "marks": 65},
    {"name": "Karim", "marks": 45}
]


def check_result(marks):

    if marks >= 80:
        return "A+"
    elif marks >= 60:
        return "B"
    else:
        return "Fail"


for student in students:

    grade = check_result(student["marks"])

    print("Name:", student["name"])
    print("Marks:", student["marks"])
    print("Grade:", grade)
    print("----------------")
```
